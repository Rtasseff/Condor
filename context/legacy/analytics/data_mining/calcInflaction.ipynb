```python
import pandas as pd

firstYYYY = 1958
lastYYYY = 2024
lastMM=9

inPath = '/Users/rtasseff/projects/tmp/CPI_BLS_2024-1958.csv'
outPath = '/Users/rtasseff/projects/tmp/CPICum_2024-1958.csv'

data = pd.read_csv(inPath)

cCpi = 1

with open(outPath,'w') as f:
    f.write('Year,Month,cCpi_2024'+str(lastMM+1))
    for YYYY in range(lastYYYY,firstYYYY-1,-1):
        for MM in range(12,0,-1):
            print(YYYY)
            print(MM)
            
            if not (YYYY==lastYYYY and MM>lastMM):
                query='Year == '+str(YYYY)+' and Period == "'+str(MM)+'"'
                print(query)
                cpi = float(data.query(query)['Value'][0])
                cCpi *= (1+cpi)
                f.write(str(YYYY)+','+str(MM)+','+str(cCpi)+'\n')
    f.close()
        


        

```
